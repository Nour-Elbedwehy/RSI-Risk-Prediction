### 8. Machine Learning Preprocessing Pipeline
Implementing a robust ColumnTransformer and preventing Data Leakage.

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder, LabelEncoder

# 1. Define Features (X) and Target (y)
X = df.drop(columns=['ProgrammerID', 'InjuryRiskLevel'])
y_raw = df['InjuryRiskLevel']

# 2. Encode the Target Variable
le = LabelEncoder()
y = le.fit_transform(y_raw)
print(f"Target classes mapped as: {dict(zip(le.classes_, le.transform(le.classes_)))}")

# 3. Train/Test Split (Crucial Data Leakage Prevention)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.20, 
    random_state=42, 
    stratify=y
)

print(f"Training data shape: {X_train.shape}")
print(f"Testing data shape: {X_test.shape}")

# 4. Define Feature Lists
numerical_features = [
    'Age', 'DailySittingHours', 'BreakCountPerDay', 
    'YearsExperience', 'SleepHours', 'TypingHoursPerDay', 'PainLevel'
]

categorical_features = [
    'Gender', 'Posture', 'KeyboardType', 
    'DeviceType', 'DoesStretching', 'JobRole'
]

# 5. Build Modular Pipelines
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')), 
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False)) 
])

# 6. Combine all transformers into a Single ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numerical_features),
        ('cat', categorical_transformer, categorical_features)
    ])

print("Pipeline Architecture Created Successfully!")
```

### 9. Baseline Model Training and Evaluation
Evaluating Logistic Regression, Random Forest, and XGBoost using our Preprocessing Pipeline.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import classification_report, accuracy_score
import time

# 1. Define the Models
models = {
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'Random Forest': RandomForestClassifier(random_state=42),
    'XGBoost': XGBClassifier(random_state=42, use_label_encoder=False, eval_metric='mlogloss')
}

# 2. Train and Evaluate each Model
model_results = {}

for name, model in models.items():
    print(f"--- Training {name} ---")
    start_time = time.time()
    
    # Combine preprocessor and model into a single pipeline
    clf = Pipeline(steps=[
        ('preprocessor', preprocessor),
        ('classifier', model)
    ])
    
    # Train the pipeline on the training data
    clf.fit(X_train, y_train)
    
    # Predict on the testing data
    y_pred = clf.predict(X_test)
    
    end_time = time.time()
    
    # Evaluate performance
    acc = accuracy_score(y_test, y_pred)
    model_results[name] = clf
    
    print(f"Accuracy: {acc:.4f}  |  Time Taken: {end_time - start_time:.2f}s")
    print(classification_report(y_test, y_pred, target_names=le.classes_))
    print("="*50)
```

### 10. Improved Machine Learning Pipeline

### Change Log

**Old implementation:** The notebook explored `df_Cleaned` and `df_prep`, but the final models used the original `df`. This meant duplicate removal and the earlier manual cleaning were not used during training. The models also used one train/test split, default settings, and accuracy plus a classification report only.

**Problems found and changes made:**

1. **Duplicate rows:** the old cleaning section removed duplicates from `df_Cleaned`, but training used `df`, so 65 duplicate rows remained. The improved pipeline removes exact duplicates before creating `X` and `y`. This prevents the same person from influencing both training and testing.
2. **Preprocessing consistency:** the old manual imputation and encoding were separate from model training. The improved pipeline uses one `ColumnTransformer` for every model. Missing values are learned from training data only, and unknown categories are handled safely.
3. **Missing values:** the old pipeline imputed values correctly, but it did not keep information about which values were missing. The numeric imputer now also creates missing-value indicators, because missingness can contain useful information.
4. **Feature quality:** the old model used only the original columns. Two domain-based workload and recovery features are added below. They are simple combinations of typing, sitting, and breaks.
5. **Class balance:** High risk is the smallest class. Suitable models use balanced class weights so that the training objective does not focus mostly on Low and Medium risk.
6. **Model comparison:** the old notebook compared three default models. The improved section keeps them and adds several simple classical models with different learning styles.
7. **Validation and tuning:** the old notebook used one split and no tuning. The improved pipeline uses stratified cross-validation with a small parameter grid for selected tree models.
8. **Evaluation:** the old notebook reported accuracy and a classification report. The improved section adds macro precision, recall, F1, train/test comparison, confusion matrix, and multiclass ROC-AUC. Macro scores give each risk class equal importance.

```python
# 1. Remove exact duplicate rows before splitting the data
df_Model = df.drop_duplicates().copy()

print("Original dataset shape:", df.shape)
print("Dataset shape after removing duplicates:", df_Model.shape)
print("Duplicate rows removed:", df.shape[0] - df_Model.shape[0])
```

### 10.1 Feature Engineering

Two simple features are added. `TypingToBreakRatio` represents typing workload relative to recovery breaks. `SittingToBreakRatio` represents sitting time relative to breaks. A value of one is added to the break count only to avoid division by zero. Both features are logical RSI-risk indicators and are still handled by the same training pipeline.

```python
# 2. Create simple workload and recovery features
df_Model["TypingToBreakRatio"] = (
    df_Model["TypingHoursPerDay"] / (df_Model["BreakCountPerDay"] + 1)
)

df_Model["SittingToBreakRatio"] = (
    df_Model["DailySittingHours"] / (df_Model["BreakCountPerDay"] + 1)
)

display(df_Model[["TypingToBreakRatio", "SittingToBreakRatio"]].head())
```

```python
from sklearn.model_selection import StratifiedKFold, GridSearchCV
from sklearn.ensemble import ExtraTreesClassifier, GradientBoostingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import precision_score, recall_score, f1_score, confusion_matrix, roc_auc_score

# 3. Define features and target after duplicate removal
X = df_Model.drop(columns=["ProgrammerID", "InjuryRiskLevel"])
y_raw = df_Model["InjuryRiskLevel"]

le = LabelEncoder()
y = le.fit_transform(y_raw)

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

numerical_features = [
    "Age", "DailySittingHours", "BreakCountPerDay",
    "YearsExperience", "SleepHours", "TypingHoursPerDay",
    "PainLevel", "TypingToBreakRatio", "SittingToBreakRatio"
]

categorical_features = [
    "Gender", "Posture", "KeyboardType",
    "DeviceType", "DoesStretching", "JobRole"
]

numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median", add_indicator=True)),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])

preprocessor = ColumnTransformer(transformers=[
    ("num", numeric_transformer, numerical_features),
    ("cat", categorical_transformer, categorical_features)
])

cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
print("Improved preprocessing pipeline created successfully")
```

### 10.2 Model Comparison

The original Logistic Regression, Random Forest, and XGBoost models are kept. Extra Trees, Gradient Boosting, Decision Tree, KNN, and Linear SVM are added because they are common, suitable classifiers for a small mixed-feature tabular dataset.

```python
# 4. Define simple models for comparison
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42, class_weight="balanced"),
    "Random Forest": RandomForestClassifier(n_estimators=200, random_state=42, class_weight="balanced"),
    "XGBoost": XGBClassifier(random_state=42, eval_metric="mlogloss"),
    "Extra Trees": ExtraTreesClassifier(n_estimators=200, random_state=42, class_weight="balanced"),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42),
    "Decision Tree": DecisionTreeClassifier(random_state=42, class_weight="balanced"),
    "KNN": KNeighborsClassifier(n_neighbors=7),
    "Linear SVM": SVC(kernel="linear", probability=True, random_state=42, class_weight="balanced")
}

model_results = {}
result_rows = []

for name, model in models.items():
    print(f"--- Training {name} ---")
    start_time = time.time()

    clf = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("classifier", model)
    ])

    clf.fit(X_train, y_train)
    y_train_pred = clf.predict(X_train)
    y_pred = clf.predict(X_test)
    end_time = time.time()

    model_results[name] = clf
    result_rows.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, average="macro", zero_division=0),
        "Recall": recall_score(y_test, y_pred, average="macro", zero_division=0),
        "F1": f1_score(y_test, y_pred, average="macro", zero_division=0),
        "Train Accuracy": accuracy_score(y_train, y_train_pred),
        "Training Time": end_time - start_time
    })

results_df = pd.DataFrame(result_rows).sort_values("F1", ascending=False)
display(results_df.style.highlight_max(subset=["F1"], color="lightgreen").format({
    "Accuracy": "{:.3f}", "Precision": "{:.3f}", "Recall": "{:.3f}",
    "F1": "{:.3f}", "Train Accuracy": "{:.3f}", "Training Time": "{:.2f}s"
}))
```

### 10.3 Simple Hyperparameter Tuning

Small parameter grids are used for three tree-based models. The cross-validation score is macro F1, so selecting parameters gives equal importance to Low, Medium, and High risk.

```python
# 5. Tune selected models with small and readable parameter grids
tuning_models = {
    "Tuned Random Forest": (
        RandomForestClassifier(random_state=42, class_weight="balanced"),
        {
            "classifier__n_estimators": [200, 300],
            "classifier__max_depth": [None, 10],
            "classifier__min_samples_leaf": [1, 3]
        }
    ),
    "Tuned Extra Trees": (
        ExtraTreesClassifier(random_state=42, class_weight="balanced"),
        {
            "classifier__n_estimators": [200, 300],
            "classifier__max_depth": [None, 10],
            "classifier__min_samples_leaf": [1, 3]
        }
    ),
    "Tuned Gradient Boosting": (
        GradientBoostingClassifier(random_state=42),
        {
            "classifier__n_estimators": [100, 200],
            "classifier__learning_rate": [0.05, 0.10],
            "classifier__max_depth": [2, 3]
        }
    )
}

for name, (model, param_grid) in tuning_models.items():
    print(f"--- Tuning {name} ---")
    clf = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("classifier", model)
    ])

    search = GridSearchCV(
        clf,
        param_grid=param_grid,
        scoring="f1_macro",
        cv=cv,
        n_jobs=1
    )

    start_time = time.time()
    search.fit(X_train, y_train)
    end_time = time.time()

    best_clf = search.best_estimator_
    y_train_pred = best_clf.predict(X_train)
    y_pred = best_clf.predict(X_test)

    model_results[name] = best_clf
    result_rows.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, average="macro", zero_division=0),
        "Recall": recall_score(y_test, y_pred, average="macro", zero_division=0),
        "F1": f1_score(y_test, y_pred, average="macro", zero_division=0),
        "Train Accuracy": accuracy_score(y_train, y_train_pred),
        "Training Time": end_time - start_time
    })

    print("Best parameters:", search.best_params_)
    print(f"Best CV Macro F1: {search.best_score_:.3f}")

results_df = pd.DataFrame(result_rows).sort_values("F1", ascending=False).reset_index(drop=True)
display(results_df.style.highlight_max(subset=["F1"], color="lightgreen").format({
    "Accuracy": "{:.3f}", "Precision": "{:.3f}", "Recall": "{:.3f}",
    "F1": "{:.3f}", "Train Accuracy": "{:.3f}", "Training Time": "{:.2f}s"
}))
```

### 10.4 Best Model

The comparison in 10.2 only covered the baseline models. The tuned models from 10.3 were added to `result_rows` but the comparison table was never rebuilt, so the best model could have been missed. This section rebuilds the full comparison table using every baseline and tuned model together, sorts it by test macro F1, and selects the true best-performing model to carry into evaluation, feature importance, and saving.

```python
# Rebuild the full comparison table using ALL models: baseline (10.2) + tuned (10.3)
full_results_df = pd.DataFrame(result_rows).sort_values("F1", ascending=False).reset_index(drop=True)

print("Full Model Comparison (Baseline + Tuned):")
display(full_results_df.style.highlight_max(subset=["F1"], color="lightgreen").format({
    "Accuracy": "{:.3f}", "Precision": "{:.3f}", "Recall": "{:.3f}",
    "F1": "{:.3f}", "Train Accuracy": "{:.3f}", "Training Time": "{:.2f}s"
}))

# Use the corrected table downstream (Final Evaluation, Feature Importance, Save Best Model)
results_df = full_results_df

best_model_name = results_df.iloc[0]["Model"]
best_model = model_results[best_model_name]

print(f"\nBest Model Selected: {best_model_name}")
print(f"Test F1 (macro):  {results_df.iloc[0]['F1']:.3f}")
print(f"Test Accuracy:    {results_df.iloc[0]['Accuracy']:.3f}")
print(f"Test Precision:   {results_df.iloc[0]['Precision']:.3f}")
print(f"Test Recall:      {results_df.iloc[0]['Recall']:.3f}")
```

### 10.5 Final Evaluation

The final comparison is sorted by test macro F1. The best displayed model is examined with a classification report, confusion matrix, ROC-AUC, and train/test performance gap.

### 10.6 ROC Curve and AUC Score

This section evaluates the best model's ability to separate the three risk classes (Low, Medium, High) using One-vs-Rest ROC curves. Each class is treated as "positive" against the rest, giving one ROC curve and one AUC score per class, plus the overall macro-average AUC already computed in 10.5.

```python
from sklearn.preprocessing import label_binarize
from sklearn.metrics import roc_curve, auc

# 1. Binarize the true labels for One-vs-Rest ROC curves (needed for multi-class problems)
classes = le.classes_
n_classes = len(classes)
y_test_binarized = label_binarize(y_test, classes=range(n_classes))

# 2. Compute the ROC curve and AUC score for each class
fpr = {}
tpr = {}
roc_auc_per_class = {}

for i in range(n_classes):
    fpr[i], tpr[i], _ = roc_curve(y_test_binarized[:, i], y_proba[:, i])
    roc_auc_per_class[i] = auc(fpr[i], tpr[i])

# 3. Print AUC per class and the overall macro-average AUC (matches test_roc_auc from 10.5)
print("AUC Score per class (One-vs-Rest):")
for i, class_name in enumerate(classes):
    print(f"  {class_name}: {roc_auc_per_class[i]:.3f}")

print(f"\nMacro-average AUC: {test_roc_auc:.3f}")

# 4. Plot the ROC curve for each class
plt.figure(figsize=(7, 6))
colors = ["darkorange", "seagreen", "royalblue"]

for i, class_name in enumerate(classes):
    plt.plot(
        fpr[i], tpr[i],
        color=colors[i % len(colors)], lw=2,
        label=f"{class_name} (AUC = {roc_auc_per_class[i]:.3f})"
    )

plt.plot([0, 1], [0, 1], "k--", lw=1, label="Chance level (AUC = 0.500)")

plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title(f"ROC Curve (One-vs-Rest) - {best_model_name}")
plt.legend(loc="lower right")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### 10.7 Feature Importance

Which features drive the model's predictions the most? The method used depends on the type of the best model:
- Tree-based models (Random Forest, XGBoost, Extra Trees, Gradient Boosting, Decision Tree) use their built-in impurity-based `feature_importances_`.
- Linear models (Logistic Regression, Linear SVM) use the absolute value of their coefficients, averaged across classes for multi-class problems.
- Any other model (e.g. KNN) falls back to Permutation Importance, which measures the drop in macro F1 score when a feature's values are shuffled.

```python
print("Feature Importance for the Best Model:", best_model_name)

# 1. Get feature names after preprocessing (numeric + one-hot encoded categorical columns)
feature_names = best_model.named_steps["preprocessor"].get_feature_names_out()

classifier = best_model.named_steps["classifier"]

# 2. Extract importances depending on the type of the best model
if hasattr(classifier, "feature_importances_"):
    importances = classifier.feature_importances_
    importance_type = "Feature Importance (impurity-based)"

elif hasattr(classifier, "coef_"):
    coefs = classifier.coef_
    # Multi-class: average the absolute coefficient across classes
    if coefs.ndim > 1:
        importances = np.abs(coefs).mean(axis=0)
    else:
        importances = np.abs(coefs)
    importance_type = "Feature Importance (|coefficient|, averaged across classes)"

else:
    # Fallback for models with neither attribute (e.g. KNN): Permutation Importance
    from sklearn.inspection import permutation_importance

    perm_result = permutation_importance(
        best_model, X_test, y_test,
        n_repeats=10, random_state=42, scoring="f1_macro", n_jobs=1
    )
    importances = perm_result.importances_mean
    importance_type = "Feature Importance (permutation, drop in macro F1)"

# 3. Build a sorted DataFrame of features and their importance
feature_importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importances
}).sort_values("Importance", ascending=False).reset_index(drop=True)

print(f"\nMethod used: {importance_type}\n")
display(feature_importance_df)

# 4. Plot the Top N most important features
top_n = min(15, len(feature_importance_df))
top_features_plot = feature_importance_df.head(top_n).iloc[::-1]

plt.figure(figsize=(9, 7))
plt.barh(top_features_plot["Feature"], top_features_plot["Importance"], color="steelblue")
plt.title(f"Top {top_n} Feature Importances - {best_model_name}")
plt.xlabel(importance_type)
plt.ylabel("Feature")
plt.grid(axis="x", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

### 12. Simple Balanced Random Forest Model

This section uses a simple manual approach to remove duplicates, handle missing values, add two workload features, and train a balanced Random Forest model.

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, f1_score

# 1. Remove exact duplicate rows
df_SimpleModel = df.drop_duplicates().copy()

print("Dataset shape after removing duplicates:", df_SimpleModel.shape)

# 2. Add simple workload and break features
df_SimpleModel["TypingToBreakRatio"] = (
    df_SimpleModel["TypingHoursPerDay"] / (df_SimpleModel["BreakCountPerDay"] + 1)
)

df_SimpleModel["SittingToBreakRatio"] = (
    df_SimpleModel["DailySittingHours"] / (df_SimpleModel["BreakCountPerDay"] + 1)
)

# This score gives more weight to sitting and pain, and less to breaks
df_SimpleModel["RiskScore"] = (
    df_SimpleModel["DailySittingHours"]
    + df_SimpleModel["PainLevel"]
    - df_SimpleModel["BreakCountPerDay"]
)

# 3. Separate features and target
X_Simple = df_SimpleModel.drop(columns=["ProgrammerID", "InjuryRiskLevel"])
y_Simple = df_SimpleModel["InjuryRiskLevel"]

# 4. Split the data before filling missing values
X_train_Simple, X_test_Simple, y_train_Simple, y_test_Simple = train_test_split(
    X_Simple,
    y_Simple,
    test_size=0.20,
    random_state=42,
    stratify=y_Simple
)

# 5. Find numeric and categorical columns
numeric_columns_Simple = X_train_Simple.select_dtypes(include="number").columns
categorical_columns_Simple = X_train_Simple.select_dtypes(exclude="number").columns

# 6. Fill missing numeric values with training medians
for column in numeric_columns_Simple:
    median_value = X_train_Simple[column].median()
    X_train_Simple[column] = X_train_Simple[column].fillna(median_value)
    X_test_Simple[column] = X_test_Simple[column].fillna(median_value)

# 7. Fill missing categorical values with training modes
for column in categorical_columns_Simple:
    mode_value = X_train_Simple[column].mode()[0]
    X_train_Simple[column] = X_train_Simple[column].fillna(mode_value)
    X_test_Simple[column] = X_test_Simple[column].fillna(mode_value)

# 8. Convert text columns into numeric dummy columns
X_train_Simple = pd.get_dummies(
    X_train_Simple,
    columns=categorical_columns_Simple,
    drop_first=True
)

X_test_Simple = pd.get_dummies(
    X_test_Simple,
    columns=categorical_columns_Simple,
    drop_first=True
)

# 9. Make sure train and test have the same columns
X_train_Simple, X_test_Simple = X_train_Simple.align(
    X_test_Simple,
    join="left",
    axis=1,
    fill_value=0
)

# 10. Encode Low, Medium, and High target values
label_encoder_Simple = LabelEncoder()
y_train_Simple_Encoded = label_encoder_Simple.fit_transform(y_train_Simple)
y_test_Simple_Encoded = label_encoder_Simple.transform(y_test_Simple)

print("Class mapping:", dict(zip(
    label_encoder_Simple.classes_,
    label_encoder_Simple.transform(label_encoder_Simple.classes_)
)))

# 11. Train a balanced Random Forest model
simple_model = RandomForestClassifier(
    n_estimators=300,
    max_depth=8,
    min_samples_leaf=3,
    class_weight="balanced",
    random_state=42
)

simple_model.fit(X_train_Simple, y_train_Simple_Encoded)

# 12. Predict the test data
y_pred_Simple = simple_model.predict(X_test_Simple)

# 13. Show the main evaluation scores
accuracy_Simple = accuracy_score(y_test_Simple_Encoded, y_pred_Simple)
macro_f1_Simple = f1_score(y_test_Simple_Encoded, y_pred_Simple, average="macro")

print("\
Accuracy:", round(accuracy_Simple, 3))
print("Macro F1 Score:", round(macro_f1_Simple, 3))

print("\
Classification Report:")
print(classification_report(
    y_test_Simple_Encoded,
    y_pred_Simple,
    target_names=label_encoder_Simple.classes_
))

# 14. Show where the risk classes are being confused
cm_Simple = confusion_matrix(y_test_Simple_Encoded, y_pred_Simple)
display_plot_Simple = ConfusionMatrixDisplay(
    confusion_matrix=cm_Simple,
    display_labels=label_encoder_Simple.classes_
)

display_plot_Simple.plot(cmap="Blues")
plt.title("Confusion Matrix - Simple Balanced Random Forest")
plt.show()

# 15. Show the most useful features according to the model
feature_importance_Simple = pd.DataFrame({
    "Feature": X_train_Simple.columns,
    "Importance": simple_model.feature_importances_
})

feature_importance_Simple = feature_importance_Simple.sort_values(
    by="Importance",
    ascending=False
)

print("\
Top 15 Important Features:")
display(feature_importance_Simple.head(15))

# 16. Train a simple balanced Logistic Regression model for comparison
from sklearn.linear_model import LogisticRegression

ord_model = LogisticRegression(
    class_weight="balanced",
    max_iter=1000,
    random_state=42
)

ord_model.fit(X_train_Simple, y_train_Simple_Encoded)
y_pred_Ord = ord_model.predict(X_test_Simple)

accuracy_Ord = accuracy_score(y_test_Simple_Encoded, y_pred_Ord)
macro_f1_Ord = f1_score(y_test_Simple_Encoded, y_pred_Ord, average="macro")

print("\nBalanced Logistic Regression Results")
print("Accuracy:", round(accuracy_Ord, 3))
print("Macro F1 Score:", round(macro_f1_Ord, 3))

# 17. Compare the two simple models
simple_results = pd.DataFrame({
    "Model": ["Balanced Random Forest", "Balanced Logistic Regression"],
    "Accuracy": [accuracy_Simple, accuracy_Ord],
    "Macro F1": [macro_f1_Simple, macro_f1_Ord]
})

simple_results = simple_results.sort_values("Macro F1", ascending=False)
print("\nSimple Model Comparison:")
display(simple_results)
```

### 13. Binary Risk Classification

This section tests two simple binary versions of the target. First, Medium is removed. Second, Medium is merged with High. These are experiments only: the final choice should match the project definition of risk.

```python
from sklearn.linear_model import LogisticRegression


def test_binary_model(df_binary, title):
    X_binary = df_binary.drop(columns=["ProgrammerID", "InjuryRiskLevel"])
    y_binary = df_binary["InjuryRiskLevel"]

    X_train_binary, X_test_binary, y_train_binary, y_test_binary = train_test_split(
        X_binary,
        y_binary,
        test_size=0.20,
        random_state=42,
        stratify=y_binary
    )

    numeric_columns_binary = X_train_binary.select_dtypes(include="number").columns
    categorical_columns_binary = X_train_binary.select_dtypes(exclude="number").columns

    # Fill missing values using only training data values
    for column in numeric_columns_binary:
        median_value = X_train_binary[column].median()
        X_train_binary[column] = X_train_binary[column].fillna(median_value)
        X_test_binary[column] = X_test_binary[column].fillna(median_value)

    for column in categorical_columns_binary:
        mode_value = X_train_binary[column].mode()[0]
        X_train_binary[column] = X_train_binary[column].fillna(mode_value)
        X_test_binary[column] = X_test_binary[column].fillna(mode_value)

    # Convert text columns to numbers
    X_train_binary = pd.get_dummies(
        X_train_binary,
        columns=categorical_columns_binary,
        drop_first=True
    )

    X_test_binary = pd.get_dummies(
        X_test_binary,
        columns=categorical_columns_binary,
        drop_first=True
    )

    X_train_binary, X_test_binary = X_train_binary.align(
        X_test_binary,
        join="left",
        axis=1,
        fill_value=0
    )

    # Train a balanced Logistic Regression model
    binary_model = LogisticRegression(
        class_weight="balanced",
        max_iter=1000,
        random_state=42
    )

    binary_model.fit(X_train_binary, y_train_binary)
    y_pred_binary = binary_model.predict(X_test_binary)

    accuracy_binary = accuracy_score(y_test_binary, y_pred_binary)
    f1_binary = f1_score(y_test_binary, y_pred_binary, average="macro")

    print(f"\n--- {title} ---")
    print("Class counts:")
    print(y_binary.value_counts())
    print("Accuracy:", round(accuracy_binary, 3))
    print("Macro F1 Score:", round(f1_binary, 3))
    print(classification_report(y_test_binary, y_pred_binary))

    return accuracy_binary, f1_binary


# 1. Remove Medium risk rows and compare Low against High
df_Binary_Remove = df_SimpleModel[
    df_SimpleModel["InjuryRiskLevel"] != "Medium"
].copy()

accuracy_remove, f1_remove = test_binary_model(
    df_Binary_Remove,
    "Binary Classification: Low vs High after removing Medium"
)


# 2. Merge Medium with High and compare Low against High-or-Medium
df_Binary_Merge = df_SimpleModel.copy()
df_Binary_Merge["InjuryRiskLevel"] = df_Binary_Merge[
    "InjuryRiskLevel"
].replace({"Medium": "High"})

accuracy_merge, f1_merge = test_binary_model(
    df_Binary_Merge,
    "Binary Classification: Low vs High after merging Medium with High"
)


# Show both binary experiments together
binary_results = pd.DataFrame({
    "Experiment": ["Remove Medium", "Merge Medium with High"],
    "Accuracy": [accuracy_remove, accuracy_merge],
    "Macro F1": [f1_remove, f1_merge]
})

binary_results = binary_results.sort_values("Macro F1", ascending=False)
print("\nBinary Classification Comparison:")
display(binary_results)
```

### 14. Simple Final Binary Model

This is a simple final version that removes Medium and predicts only Low or High risk.

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, ConfusionMatrixDisplay, f1_score
import matplotlib.pyplot as plt
import pandas as pd

# 1. Copy the original data and remove duplicate rows
data_for_binary_model = df.drop_duplicates().copy()

# 2. Remove Medium risk rows
data_for_binary_model = data_for_binary_model[
    data_for_binary_model["InjuryRiskLevel"] != "Medium"
].copy()

print("Data shape after removing Medium:", data_for_binary_model.shape)
print(data_for_binary_model["InjuryRiskLevel"].value_counts())

# 3. Create one simple risk feature
data_for_binary_model["RiskScore"] = (
    data_for_binary_model["DailySittingHours"]
    + data_for_binary_model["PainLevel"]
    - data_for_binary_model["BreakCountPerDay"]
)

# 4. Separate input columns from the risk label
features_for_binary_model = data_for_binary_model.drop(
    columns=["ProgrammerID", "InjuryRiskLevel"]
)
risk_labels_for_binary_model = data_for_binary_model["InjuryRiskLevel"]

# 5. Split the data into training data and test data
training_features, testing_features, training_labels, testing_labels = train_test_split(
    features_for_binary_model,
    risk_labels_for_binary_model,
    test_size=0.20,
    random_state=42,
    stratify=risk_labels_for_binary_model
)

# 6. Encode labels
label_encoder = LabelEncoder()
y_train_encoded = label_encoder.fit_transform(training_labels)
y_test_encoded = label_encoder.transform(testing_labels)

# 7. Identify numeric and categorical columns
numeric_columns = training_features.select_dtypes(include="number").columns.tolist()
categorical_columns = training_features.select_dtypes(exclude="number").columns.tolist()

# 8. Build a clean preprocessing pipeline
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_columns),
        ("cat", categorical_transformer, categorical_columns)
    ]
)

# 9. Train a balanced Logistic Regression model within a clean Pipeline
final_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        class_weight="balanced",
        max_iter=1000,
        random_state=42
    ))
])

final_model.fit(training_features, y_train_encoded)

# Define X_test for subsequent prediction cell
X_test = testing_features

# 10. Predict Low risk or High risk on the test data
predicted_encoded_labels = final_model.predict(X_test)
predicted_risk_labels = label_encoder.inverse_transform(predicted_encoded_labels)

# 11. Show the final results
binary_accuracy = accuracy_score(testing_labels, predicted_risk_labels)
binary_macro_f1 = f1_score(testing_labels, predicted_risk_labels, average="macro")

print("\nFinal Binary Model Results")
print("Accuracy:", round(binary_accuracy, 3))
print("Macro F1 Score:", round(binary_macro_f1, 3))

print("\nClassification Report:")
print(classification_report(testing_labels, predicted_risk_labels))

# 12. Show the prediction mistakes
binary_confusion_matrix = confusion_matrix(
    testing_labels,
    predicted_risk_labels,
    labels=label_encoder.classes_
)

binary_confusion_matrix_display = ConfusionMatrixDisplay(
    confusion_matrix=binary_confusion_matrix,
    display_labels=label_encoder.classes_
)

binary_confusion_matrix_display.plot(cmap="Blues")
plt.title("Confusion Matrix - Low and High Risk")
plt.show()
```
